```python
import numpy as np
import pandas as pd

from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv("NABIL.csv")

data["published_date"] = pd.to_datetime(
    data["published_date"]
).dt.normalize()

data["close"] = pd.to_numeric(
    data["close"], errors="coerce"
)

data = data.dropna(subset=["published_date", "close"])
data = data[data["close"] > 0].copy()

# Repeated dates should have the same closing price.
different_prices = (
    data.groupby("published_date")["close"].nunique()
)

if (different_prices > 1).any():
    raise ValueError("Some repeated dates have different closing prices.")

data = data.drop_duplicates(subset="published_date")

data = data.sort_values("published_date")
data = data.reset_index(drop=True)

print("Total trading sessions:", len(data))
print("First date:", data["published_date"].iloc[0].date())
print("Last date:", data["published_date"].iloc[-1].date())

data.head()
```

```python
LOOKBACK = 15

features = [
    f"close_{day}"
    for day in range(1, LOOKBACK + 1)
]

samples = data[["published_date", "close"]].copy()

for day in range(1, LOOKBACK + 1):
    samples[f"close_{day}"] = samples["close"].shift(day)

samples = samples.rename(columns={"close": "actual_close"})

samples = samples.dropna(
    subset=features
).reset_index(drop=True)

samples[
    ["published_date", "actual_close", "close_1", "close_15"]
].head()
```

```python
def predict_date(date_text):
    try:
        chosen_date = pd.Timestamp(date_text).normalize()
    except (ValueError, TypeError):
        print("Enter a valid date, such as 2026-09-23.")
        return

    if pd.isna(chosen_date):
        print("Please enter a date.")
        return

    selected = samples[
        samples["published_date"] == chosen_date
    ]

    if selected.empty:
        if chosen_date > data["published_date"].max():
            print("That date is beyond the data in your CSV.")
            print("Update the CSV to check its actual closing price.")
        elif chosen_date in data["published_date"].values:
            print("There are fewer than 15 earlier sessions for that date.")
        else:
            print("That date is not recorded in your CSV.")
            print("Choose a recorded trading date.")
        return

    # Keep the chosen date and all later dates out of training.
    train = samples[
        samples["published_date"] < chosen_date
    ]

    if len(train) < 30:
        print("Choose a later date so there are at least 30 training examples.")
        return

    model = make_pipeline(
        StandardScaler(),
        Ridge(alpha=10.0)
    )

    model.fit(
        train[features],
        train["actual_close"]
    )

    predicted_price = model.predict(
        selected[features]
    )[0]

    actual_price = selected["actual_close"].iloc[0]
    baseline_price = selected["close_1"].iloc[0]

    model_error = abs(actual_price - predicted_price)
    baseline_error = abs(actual_price - baseline_price)

    previous_days = data[
        data["published_date"] < chosen_date
    ].tail(LOOKBACK)

    print("Prediction date:", chosen_date.date())
    print("Training examples:", len(train))
    print(
        "15-session input window:",
        previous_days["published_date"].iloc[0].date(),
        "to",
        previous_days["published_date"].iloc[-1].date()
    )

    print(f"\nPredicted closing price: Rs. {predicted_price:.2f}")
    print(f"Actual closing price:    Rs. {actual_price:.2f}")
    print(f"Absolute error:          Rs. {model_error:.2f}")

    print(f"\nPrevious-close baseline: Rs. {baseline_price:.2f}")
    print(f"Baseline absolute error: Rs. {baseline_error:.2f}")

    if model_error < baseline_error:
        print("\nRidge was more accurate on this date.")
    elif model_error > baseline_error:
        print("\nThe baseline was more accurate on this date.")
    else:
        print("\nBoth predictions had the same error.")

    print("\nClosing prices used as inputs:")
    display(previous_days[["published_date", "close"]])

    return pd.DataFrame({
        "date": [chosen_date],
        "predicted_close": [predicted_price],
        "actual_close": [actual_price],
        "absolute_error": [model_error],
        "baseline_close": [baseline_price],
        "baseline_error": [baseline_error]
    })
```

```python
date_text = input("Enter prediction date (YYYY-MM-DD): ").strip()

result = predict_date(date_text)

if result is not None:
    display(result)
    
```

```python
if result is not None:
    predicted_price = result["predicted_close"].iloc[0]
    actual_price = result["actual_close"].iloc[0]
    previous_close = result["baseline_close"].iloc[0]

    def direction(price, previous_price):
        if price > previous_price:
            return "UP"
        elif price < previous_price:
            return "DOWN"
        else:
            return "UNCHANGED"

    predicted_direction = direction(predicted_price, previous_close)
    actual_direction = direction(actual_price, previous_close)

    # 0 = correct direction, 1 = incorrect direction
    direction_error = int(predicted_direction != actual_direction)

    print("Model: Ridge regression")
    print("Predicted direction:", predicted_direction)
    print("Actual direction:", actual_direction)
    print("Direction error:", direction_error)

else:
    print("Run the prediction cell and choose a valid date first.")
```
